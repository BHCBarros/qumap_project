# Summary

This project aims to simulate the algorithm presented in the paper *Tackling the Qubit Mapping Problem for NISQ-Era Quantum Devices* published in ASPLOS 19 being cited as:

Gushu Li, Yufei Ding, Yuan Xie. **Tackling the Qubit Mapping Problem for NISQ-Era Quantum Devices**, ASPLOS 19: Proceedings of The Twenty-Fourth International Conference on Architectural Support for Programming Languages and Operating Systems. Pages 1001 - 1014 (2019). https://doi.org/10.1145/3297858.3304023

As we enter an era where physical systems can contain many qubits ("many" here is compared with the quantum systems that came before it; if comparing with classical systems, the amount of qubits implemented is still small), the NISQ (Noisy Intermediate Scale Quantum) devices era, the problem of the connectivity of these qubits becomes of significant importance. In a real quantum chip not every qubit is be directly connected to each other and that needs to be taken into consideration when designing quantum circuits to run in such chips. The authors here propose an algorithm to translate an arbitrary circuit created without specific concearns of the physical architecture that will run it into a code that respects its restraints. This project has as its goal to simulate this code using the Python library qiskit and for that the following objectives has been stablished:

1 - To implement a python code that receives a quantum circuit and a graph $G$ that tells how the qubits in the circuit are connected and outputs a new quantum circuit that is effectively the same as the input but with the constraints in $G$ respected. In this function a decay rate parameter will also be provided that will control the tradeoff between the final circuit's depth and gate count;

2 - To measure the output circuit's depth and gate count against the input circuit as well as runtime efficiency in finding the best possible gate combinations and;

3 - Apply the code to an example circuit and verify they still produce the same output.
